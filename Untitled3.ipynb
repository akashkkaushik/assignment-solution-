#a) Key differences between relational databases and MongoDB:

Data Structure: Relational databases (SQL) store data in rigid, tabular formats (rows and columns). MongoDB (NoSQL) stores data in flexible, JSON-like documents (BSON).

Scaling: SQL scales vertically (requiring a more powerful CPU/RAM on a single server). MongoDB is designed to scale horizontally (distributing data across multiple cheaper commodity servers via sharding).

Schema: SQL requires a predefined schema before inserting data. MongoDB is dynamically typed and schema-less; documents in the same collection can have entirely different fields.

#b) When MongoDB is preferred over SQL systems:
MongoDB is preferred when building rapid prototypes, handling vast amounts of unstructured or semi-structured data (like IoT sensor logs or social media feeds), or when an application requires massive write loads that would bottleneck a traditional SQL database.

#c) Schema flexibility and its risks:
Schema flexibility allows developers to iterate quickly without running slow database migrations. However, the risk is Data Inconsistency. Because the database does not enforce strict rules, application-level code must handle missing fields, mismatched data types, and validation. If the application code has bugs, "dirty" data will silently enter the database.

#d) Embedding vs Referencing impact on performance:

Embedding (Nesting data inside a single document): Massively improves read performance because all related data is retrieved in a single disk read operation. However, it can duplicate data and cause documents to hit the 16MB BSON size limit.

Referencing (Linking documents via ObjectIDs): Keeps data normalized and avoids duplication (saving storage). However, it requires multiple queries or the $lookup aggregation to join data, which negatively impacts read performance.

#Question 2: Indexing, Sharding & Replication


a) How do indexes improve MongoDB query performance?


b) What is sharding? When is it required?


c) What is replication and how does it ensure high availability?


d) How does improper shard key selection impact performance?

#a) How indexes improve MongoDB query performance:
Without an index, MongoDB must perform a Collection Scan, reading every single document to find a match. An index organizes specific fields into a highly efficient B-Tree data structure, allowing the database engine to find target documents in logarithmic time, vastly reducing CPU and disk I/O usage.

#b) What is sharding and when is it required?
Sharding is the process of distributing data horizontally across multiple machines. It becomes required when a single machine's RAM, Disk I/O, or Storage Capacity is no longer sufficient to handle the active dataset and incoming read/write throughput.

#c) What is replication and how does it ensure high availability?
Replication is the process of synchronizing data across multiple servers (a Replica Set containing a Primary node and multiple Secondary nodes). If the Primary server crashes, the system automatically elects a Secondary node to become the new Primary. This ensures zero-downtime fault tolerance (High Availability).

#d) Impact of improper shard key selection:
An improper shard key can cause:

1)  Jumbo Chunks: Data cannot be split further, breaking horizontal distribution.



2)  Hotspots: All new writes route to a single shard instead of being evenly distributed, defeating the purpose of scaling.



3) Scatter-Gather Queries: If queries do not include the shard key, MongoDB must broadcast the query to all shards and merge the results, destroying read performance.

#Question 3: Descriptive Statistics & Distributions


a) Difference between mean, median, and variance


b) What does high variance indicate in customer spending?


c) Difference between normal distribution and skewed distribution


d) Why do business metrics often show right-skewed distributions?

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure clean notebook plotting
%matplotlib inline
sns.set_theme(style="whitegrid")

np.random.seed(42)
# Generate a Normal Distribution
normal_data = np.random.normal(loc=50, scale=10, size=1000)
# Generate a Right-Skewed Distribution (Log-Normal)
skewed_data = np.random.lognormal(mean=3.5, sigma=0.8, size=1000)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot Normal Distribution
sns.histplot(normal_data, kde=True, color="blue", ax=axes[0])
axes[0].set_title("Normal Distribution (Symmetrical)", fontsize=14)

# Plot Right-Skewed Distribution
sns.histplot(skewed_data, kde=True, color="red", ax=axes[1])
axes[1].set_title("Right-Skewed Distribution (Business Metrics)", fontsize=14)

plt.tight_layout()
plt.show()
```

#a) Difference between mean, median, and variance:

Mean: The arithmetic average of all data points.

Median: The exact middle value when the data is sorted. It is immune to extreme outliers.

Variance: A measure of dispersion showing how far, on average, the data points spread out from the mean.

#b) What high variance indicates in customer spending:
High variance means customer spending is wildly unpredictable or segmented. It indicates that you have distinct buyer personas—some customers are spending literal pennies, while "whales" are spending massive fortunes, rather than a uniform average spend.

#c) Normal vs Skewed distribution:
A normal distribution is symmetrical (bell-shaped) where the mean and median are equal. A skewed distribution is asymmetrical, pulling the "tail" and the mean toward the extreme values.

#d) Why business metrics show right-skewed distributions:
Metrics like user revenue, website visits, and sales follow the Pareto Principle (80/20 rule). A vast majority of users interact or spend very little (forming a massive peak on the left), while a tiny fraction of highly engaged users spend vast amounts (creating a long "tail" trailing off to the right).

#Question 4: Random Variables & Bayes


a) What is a random variable?


b) Difference between discrete and continuous random variables


c) Explain covariance and its interpretation


d) Explain Bayes’ theorem in business decision-making context

```python
import numpy as np
import pandas as pd

# Simulating data: Marketing Spend vs Revenue
np.random.seed(10)
marketing_spend = np.random.normal(1000, 200, 50)

# Revenue is derived from spend + some random noise (indicating positive covariance)
revenue = (marketing_spend * 3.5) + np.random.normal(0, 300, 50)

df_business = pd.DataFrame({'Marketing_Spend': marketing_spend, 'Revenue': revenue})

# Calculate Covariance Matrix
cov_matrix = np.cov(df_business['Marketing_Spend'], df_business['Revenue'])
print(f"Covariance Matrix:\n{cov_matrix}\n")
print(f"Covariance between Spend and Revenue: {cov_matrix[0, 1]:.2f}")
print("A high positive covariance confirms that as marketing spend increases, revenue increases.")
```

#a) What is a random variable?
A random variable is a mathematical formalization of a quantity whose outcomes depend on chance. It assigns a numerical value to each possible outcome of a random phenomenon.

#b) Discrete vs Continuous random variables:

   Discrete: Takes on distinct, countable values (e.g., Number of items in a shopping cart, number of customer service calls).

Continuous: Takes on an infinite number of possible values within a given range (e.g., Exact time spent on a webpage, physical weight of a package).

#c) Covariance and its interpretation:
Covariance measures the directional relationship between two random variables.

A Positive covariance means both variables tend to move in the same direction (e.g., Ad spend and Sales).

A Negative covariance means they move in opposite directions (e.g., Price increases and Sales volume).

#d) Bayes' Theorem in business decision-making:
Bayes' Theorem mathematically updates the probability of a hypothesis as new evidence becomes available. In business, it is used to pivot strategies dynamically. For example, knowing the baseline probability a user will churn, and then updating that probability the moment the user files a negative support ticket (the new evidence).

#Question 5: Write MongoDB queries to:
1. Insert a new order
2. Find all orders above ₹5000
3. Update status to "Delivered" where order_date is older than 7 days
4. Delete cancelled orders

```python
# Install mongomock if not present in the environment
!pip install mongomock -q

import mongomock
from datetime import datetime, timedelta

try:
    # Initialize mock MongoDB client
    client = mongomock.MongoClient()
    db = client.ecommerce_db
    orders = db.orders

    # Setup: Generate base synthetic data
    nine_days_ago = datetime.now() - timedelta(days=9)
    recent_date = datetime.now() - timedelta(days=2)

    orders.insert_many([
        {"order_id": 1, "amount": 6000, "status": "Processing", "order_date": nine_days_ago},
        {"order_id": 2, "amount": 3000, "status": "cancelled", "order_date": recent_date},
        {"order_id": 3, "amount": 8500, "status": "Shipped", "order_date": recent_date},
        {"order_id": 4, "amount": 4000, "status": "cancelled", "order_date": nine_days_ago},
    ])

    # --- TASK 1: Insert a new order ---
    orders.insert_one({"order_id": 5, "amount": 7500, "status": "Pending", "order_date": datetime.now()})
    print("1. Successfully inserted new order.")

    # --- TASK 2: Find all orders above ₹5000 ---
    high_value_orders = list(orders.find({"amount": {"$gt": 5000}}))
    print("\n2. Orders above ₹5000:")
    for order in high_value_orders:
        print(f" - Order {order['order_id']} | Amount: {order['amount']}")

    # --- TASK 3: Update status to 'Delivered' where order_date is older than 7 days ---
    seven_days_ago = datetime.now() - timedelta(days=7)
    update_result = orders.update_many(
        {"order_date": {"$lt": seven_days_ago}},
        {"$set": {"status": "Delivered"}}
    )
    print(f"\n3. Updated {update_result.modified_count} orders to 'Delivered'.")

    # --- TASK 4: Delete cancelled orders ---
    delete_result = orders.delete_many({"status": "cancelled"})
    print(f"\n4. Deleted {delete_result.deleted_count} cancelled orders.")

    # Final verification
    print("\nFinal Database State:")
    for doc in orders.find({}, {"_id": 0}):
        print(doc)

except Exception as e:
    print(f"MongoDB Operation Failed: {e}")
```

#1. Data Structure:

      A) Customers Collection: Store basic profile data.

      B)Products Collection: Store product catalogs.

     C)Orders Collection: Store transactional data referencing customer_id and an array of product_ids bought in that session.

#2. Aggregation Queries:
To find "Frequently bought together items", you would use the following pipeline on the Orders collection:

    $match: Filter for orders containing the target Product A.

    $unwind: Deconstruct the product_ids array into separate documents.

    $match: Filter out Product A from the unwound list (so we don't count the item against itself).

    $group: Group by product_id and use $sum: 1 to count co-occurrences.

    $sort: Sort by the count descending.

#3. How Covariance identifies relationships:
Calculating the covariance matrix across product sales velocities reveals underlying buying patterns. If Product X (Cameras) and Product Y (SD Cards) have a highly positive covariance, they are statistically bought together. If they have negative covariance, they behave as substitute goods.

#4. Sharding impact on recommendation performance:
Finding global frequently bought together items across a sharded cluster triggers a scatter-gather query. If the database is sharded by customer_id, the aggregation must run across every single shard server independently, pulling heavy amounts of data over the network to the router (mongos) for final sorting. This is highly unoptimized for real-time. Instead, these analytical workloads should be pushed to an offline processing engine (like Apache Spark) or a dedicated analytics MongoDB node.

#Question 8: Suppose:


● 5% of customers churn


● 70% of churned customers had declining order frequency


● 15% of non-churned customers had declining order frequency


If a customer shows declining frequency, compute probability that they will churn.

```python
# Known Probabilities
P_Churn = 0.05                             # P(Churn)
P_Not_Churn = 1 - P_Churn                  # P(Not Churn)
P_Decline_given_Churn = 0.70               # P(Decline | Churn)
P_Decline_given_Not_Churn = 0.15           # P(Decline | Not Churn)

# Step 1: Calculate Total Probability of Declining Order Frequency (Law of Total Probability)
# P(Decline) = [P(Decline | Churn) * P(Churn)] + [P(Decline | Not Churn) * P(Not_Churn)]
P_Decline = (P_Decline_given_Churn * P_Churn) + (P_Decline_given_Not_Churn * P_Not_Churn)

# Step 2: Apply Bayes' Theorem
# P(Churn | Decline) = [P(Decline | Churn) * P(Churn)] / P(Decline)
P_Churn_given_Decline = (P_Decline_given_Churn * P_Churn) / P_Decline

print(f"Probability a customer will churn given declining frequency: {P_Churn_given_Decline:.4f} or {P_Churn_given_Decline * 100:.2f}%")
```

#Question 9:


Explain:


1. Why CLT allows us to use normal approximation


2. Why it is important in A/B testing


3. What happens if sample size is very small?

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Simulate CLT
np.random.seed(42)

# Create a highly non-normal (Uniform) population
population = np.random.uniform(0, 100, 10000)

# Take 1000 samples, each of size n=50, and calculate their means
sample_means = [np.mean(np.random.choice(population, size=50)) for _ in range(1000)]

plt.figure(figsize=(8, 4))
sns.histplot(sample_means, kde=True, color="purple")
plt.title("Distribution of Sample Means (n=50) -> Approaches Normal")
plt.xlabel("Sample Mean")
plt.show()
```

#1. Why CLT allows us to use normal approximation:
The Central Limit Theorem states that the sampling distribution of the sample mean approaches a normal distribution as the sample size grows (typically 30 or more), regardless of the original shape of the population distribution.

#2. Why it is important in A/B testing:
In A/B testing, raw user metrics (like click-through rates or revenue per user) are heavily skewed and not normally distributed. However, because of the CLT, the average metric across thousands of users acts normally. This allows Data Scientists to use standard, highly reliable parametric statistical tests (like t-tests or Z-tests) to confidently declare a winning variant.

#3. What happens if sample size is very small?
If the sample size is small (e.g., under 30), the CLT breaks down. The sample mean distribution retains the skewness of the parent population. If you force a normal-based t-test on this data, your p-values will be wildly inaccurate. You must instead rely on non-parametric tests (like the Mann-Whitney U test).

#Question 10:The company claims:


“Average monthly spending is ₹5000.”


1. How would you statistically test this claim?


2. Define null and alternative hypothesis


3. What type of error is more dangerous in business — Type I or Type II?

```python
from scipy import stats
import numpy as np

# Synthetic sample data: 100 customers' monthly spend (True average is around 4850)
np.random.seed(0)
sample_spend = np.random.normal(loc=4850, scale=800, size=100)

claimed_mean = 5000

# 1. Statistically test the claim using a 1-Sample T-Test
t_stat, p_value = stats.ttest_1samp(sample_spend, claimed_mean)

print(f"T-Statistic: {t_stat:.2f}")
print(f"P-Value: {p_value:.5f}")

if p_value < 0.05:
    print("Conclusion: Reject the Null Hypothesis. The average is significantly different from ₹5000.")
else:
    print("Conclusion: Fail to Reject the Null Hypothesis. The claim holds.")
```

#2. Define null and alternative hypothesis:

    Null Hypothesis: The average monthly spending is exactly ₹5000. Any deviation in our sample is purely due to random chance.

    Alternative Hypothesis: The average monthly spending is NOT ₹5000. The company's claim is mathematically false.

#3. Type I vs Type II Errors in Business:

    Type I Error (False Positive): Concluding the average spend changed when it actually didn't. You might waste resources investigating a non-existent issue.

    Type II Error (False Negative): Failing to detect that the average spend has dropped when it actually has.

#Which is more dangerous?


 In this context, a Type II Error is far more dangerous. If customer spending has actually crashed below ₹5000, but your statistical test misses it (False Negative), the business will hemorrhage revenue without management ever realizing they need to deploy emergency marketing or retention strategies.
