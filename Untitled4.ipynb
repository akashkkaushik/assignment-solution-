#Question 1:Explain the complete ML lifecycle.

How is it different from traditional Software
Development Lifecycle (SDLC)?

#The Machine Learning Lifecycle consists of several iterative phases:

Problem Definition: Understanding the business goal and mapping it to an ML problem (e.g., classification, regression).



Data Collection: Gathering relevant historical data from various sources (databases, APIs, logs).



Data Preparation & Cleaning: Handling missing values, removing duplicates, and addressing outliers.



Exploratory Data Analysis (EDA): Visualizing data to understand distributions and feature correlations.



Feature Engineering: Creating new variables or transforming existing ones to help the model learn better.



Model Training: Selecting algorithms and feeding them the prepared data so they can learn patterns.



Model Evaluation & Tuning: Testing the model on unseen data using metrics (like F1-score or RMSE) and tuning hyperparameters.



Deployment: Pushing the trained model into a production environment where it can make real-time or batch predictions.



Monitoring & Maintenance: Continuously tracking the model's performance in the real world and retraining it as data changes.



#How it differs from traditional SDLC:



Deterministic vs. Probabilistic: In traditional SDLC, a developer writes explicit rules (Code + Data = Output). If the rules are correct, the output is guaranteed. In ML, the developer feeds data and outcomes into an algorithm to learn the rules (Data + Output = Code/Model). ML outputs are probabilistic (e.g., "90% confident this is a cat").



Decay: Traditional software only breaks if the environment or dependencies change. ML models "decay" inherently over time because the real-world data they encounter shifts away from the data they were trained on (concept/data drift).



Testing: SDLC uses unit and integration tests for logical bugs. ML requires statistical evaluation to ensure the model generalizes to new data without bias or overfitting.

#Question 2: Why is translating business needs into a data science problem the most critical
step?


Give one example where poor problem framing leads to wasted resources.

ranslating business needs correctly is critical because Machine Learning algorithms are literal; they will ruthlessly optimize for the exact mathematical metric you give them. If that metric does not perfectly align with the actual business objective, you will build a highly accurate model that is completely useless to the company.

1)  Example of poor problem framing:
A telecom company wants to reduce customer churn (people canceling their subscriptions). The data science team frames this as a standard classification problem and optimizes the model for Accuracy.
Because only 2% of customers churn, the model learns a lazy shortcut: it simply predicts "No one will churn" for every single customer.

The Result: The model achieves 98% accuracy (which looks great on paper), but it completely fails the business objective because it didn't identify a single churning customer. The company wasted months of salaries and compute resources. They should have framed the problem to optimize for Recall (finding all potential churners) and integrated it with a business strategy (e.g., offering a discount to predicted churners).

#Question 3:  Differentiate between:


● Predictive ML models


● Generative AI models


Give 3 real-world use cases for each.

#Predictive ML Models:
These models analyze historical data to find patterns and map inputs to specific, existing categories or numerical values. They are discriminative (they analyze and classify what already exists).



    Use Case 1: Fraud Detection (predicting if a credit card swipe is fraudulent or legitimate).

    Use Case 2: House Price Estimation (predicting the numerical sale value of a home based on square footage and location).

    Use Case 3: Medical Diagnosis (predicting the likelihood of a patient having a specific disease based on lab results).

#Generative AI Models:
These models learn the underlying distribution of a dataset in order to create entirely new, synthetic data instances (text, images, audio) that resemble the training data.

    Use Case 1: Automated Content Creation (e.g., ChatGPT writing marketing emails or summarizing long reports).

    Use Case 2: Software Code Generation (e.g., GitHub Copilot writing boilerplate Python code based on a prompt).

    Use Case 3: Image Synthesis (e.g., Midjourney generating commercial art or synthetic medical scans for training other models).

#Question 5:  

Dataset contains:


● 20% missing values


● Duplicate rows


● Imbalanced class distribution


# Explain step-by-step how you would clean and prepare this dataset.

1)   Remove Duplicate Rows: I would identify and drop exact duplicate rows first. Duplicates can skew the model's understanding of the data distribution and cause data leakage if copies of the same row end up in both the training and testing sets.

2)   Handle Missing Values (20%):
Since 20% is a significant portion, dropping these rows entirely would result in a massive loss of data.

3)   For numerical columns, I would use statistical imputation (filling missing values with the median or mean) or use advanced methods like KNN (K-Nearest Neighbors) imputation.

4)   For categorical columns, I would impute using the mode (most frequent category) or create a new category simply called "Unknown" so the model can learn if the "missingness" itself is a predictive signal.

5)   Handle Imbalanced Class Distribution:
Before addressing this, I must split the data into Train and Test sets to prevent data leakage. Once split, I will balance only the Training set.

I would use a technique like SMOTE (Synthetic Minority Over-sampling Technique) to synthetically generate new examples of the minority class, or use Random Undersampling to reduce the majority class. Alternatively, I would pass class_weights="balanced" into the ML algorithm so it mathematically penalizes errors made on the minority class.

#Question 6:

Explain:


● Tokenization

● Embeddings


● Context window


Why are embeddings critical for GenAI systems?

#Tokenization:

 The process of breaking raw text down into smaller, manageable pieces called tokens (which can be whole words, sub-words, or individual characters). For example, "Unhappy" might be tokenized into "Un" and "happy".

#Embeddings:

 A mathematical representation of tokens as dense vectors (lists of numbers) in a high-dimensional space. Words with similar semantic meanings are placed closer together in this space (e.g., the vector for "Dog" will be very close to the vector for "Puppy").

#Context Window:

 The maximum limit of tokens (input + output) a Large Language Model can process in a single interaction. If a conversation exceeds the context window, the model "forgets" the earliest parts of the prompt.

#Why embeddings are critical:

 Computers cannot understand human language; they only understand numbers. Embeddings translate semantic meaning, context, and relationships (like analogies) into complex mathematical coordinates. Without embeddings, a GenAI model would just treat words as exact keyword matches and would have zero understanding of nuance, synonyms, or language structure.

#Question 7: Explain:


● Train/Validation/Test split


● Cross-validation


● Data leakage


Why is data leakage dangerous?

#Train/Validation/Test split:

 Dividing a dataset into three parts. The Train set (e.g., 70%) is used to teach the model. The Validation set (e.g., 15%) is used to tune hyperparameters and check for overfitting during training. The Test set (e.g., 15%) is locked away until the very end to evaluate the final model's real-world performance.

#Cross-validation:

 A resampling procedure (like k-fold) used when data is limited. The data is split into 'k' chunks. The model trains on k-1 chunks and validates on the remaining 1 chunk. This repeats 'k' times so every chunk serves as validation data once, providing a more robust performance metric.

#Data Leakage:

 This occurs when information from outside the training dataset (such as from the Test set or future data) accidentally leaks into the model during the training phase.

#Why Data Leakage is Dangerous:
It gives the model an unfair advantage (like giving a student the answer key before a test). The model will show near-perfect performance during training and evaluation, tricking stakeholders into deploying it. However, once deployed in production on truly unseen data, it will fail catastrophically.

Question 8:Explain and compare:


● Accuracy


● Precision


● Recall


● F1-score


● ROC-AUC


Why is accuracy misleading in fraud detection?

#Accuracy:

 The ratio of correctly predicted observations to total observations (Total Correct / Total Predictions).

#Precision:

 Out of all the positive predictions the model made, how many were actually correct? (True Positives / (True Positives + False Positives)). High precision means fewer false alarms.

#Recall (Sensitivity):

 Out of all the actual positive cases in the dataset, how many did the model find? (True Positives / (True Positives + False Negatives)). High recall means fewer missed cases.

#F1-score:

 The harmonic mean of Precision and Recall. It provides a single metric that balances both, which is highly useful when dealing with imbalanced datasets.

#ROC-AUC:

Measures the 2D area underneath the Receiver Operating Characteristic curve. It represents the model's ability to distinguish between the positive and negative classes across all possible decision thresholds.

#Why Accuracy is Misleading in Fraud Detection:
Fraud datasets are highly imbalanced (e.g., 99.9% of transactions are legitimate; 0.1% are fraud). A completely "dumb" model that predicts every single transaction as "Legitimate" will be 99.9% accurate. Relying on accuracy masks the fact that the model caught 0% of the actual fraud. In fraud, we must look at Recall (catching the fraud) and Precision (not freezing innocent customers' cards).

Question 9:


After 6 months:


● Model accuracy drops.


● Customer behavior changes.


Explain:
1. Data drift
2. Concept drift
3. Monitoring techniques
4. Retraining strategy

#1. Data Drift (Feature Drift):

 The statistical distribution of the input data (X) has changed over time, even if the relationship with the target hasn't. Example: An inflation spike causes the average transaction amount of your customers to double compared to 6 months ago.

#2. Concept Drift:

The relationship between the input data (X) and the target variable (Y) has fundamentally changed. Example: Six months ago, a customer buying 5 laptops meant they were a business owner. Today, buying 5 laptops is a new pattern utilized by credit card fraudsters. What was once "normal" is now "fraud."

#3. Monitoring Techniques:

 * Track statistical divergence of incoming data against the baseline training data using metrics like Kullback-Leibler (KL) divergence or Population Stability Index (PSI).

    Monitor direct business KPIs (e.g., if the fraud dispute rate suddenly spikes, the model is failing).

#4. Retraining Strategy:

Implement a Continuous Training (CT) pipeline. When monitoring alerts indicate drift has exceeded an acceptable threshold, automatically trigger a retraining cycle using a sliding window of the most recent, fresh data (e.g., the last 3 months) to update the model's understanding of current behavior.

Question 10: An organization wants to integrate:


● ML fraud detection


● GenAI automated email responses


Design a high-level system architecture

explaining:
1. Data flow
2. Model interaction
3. Human-in-the-loop
4. Monitoring checkpoints

#1. Data Flow:

#Event Ingestion:

 A customer swipes their credit card. The transaction data (amount, location, time, user ID) hits the company's API Gateway.

#Predictive Inference (ML):

 The API routes the transaction data to the deployed ML Fraud Detection Model. The model evaluates the transaction in milliseconds and returns a "Fraud Score" (e.g., 92% probability of fraud).

#Decision Logic:

The backend system checks the score. If the score is > 85%, the transaction is declined, and the system flags the event for a GenAI alert.

#Context Gathering:

The backend pulls the customer's profile (name, last 3 safe locations, preferred language) from the central database.

#Generative Inference (GenAI):

The backend constructs a prompt: "Write a polite security alert to [Customer Name] explaining that a transaction of [Amount] at [Location] was blocked due to suspected fraud. Ask them to confirm if it was them." This is sent to the LLM API.

#Action:

The LLM returns a personalized, empathetic email text. The Notification Service sends this email to the customer.

#2. Model Interaction:
In this architecture, the two models act in a Trigger-and-Augment relationship.

The Predictive Model acts as the analytical brain and the gatekeeper. It processes structured numbers quickly and securely to make a binary/probabilistic decision.

#The Generative Model is subordinate;

it is only invoked based on the output of the Predictive Model. It uses the analytical decision as its context to generate unstructured, human-friendly communication, bridging the gap between a cold statistical flag and a good customer experience.
