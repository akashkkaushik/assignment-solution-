Why is FastAPI preferred over Flask for modern APIs?

#A REST API (Representational State Transfer Application Programming Interface)

 is an architectural style for designing networked applications. It relies on a stateless, client-server communications protocol—almost exclusively HTTP—using standard operations to manipulate resources identified by URIs.

#HTTP Methods ComparisonGET:

 Retrieves a representation of a resource without producing side effects on the server. It is both safe (does not modify state) and idempotent (repeated identical requests return the same result).

#POST:

 Submits data to be processed to a specified resource, typically resulting in the creation of a new subordinate resource or triggering a server-side action. It is neither safe nor idempotent.

#PUT:

 Replaces the current representation of the target resource with the uploaded request payload, or creates the resource if it does not exist. It is idempotent because sending the same replacement update multiple times produces the identical end state.

#DELETE:

 Removes the specified resource identified by the URI. It is idempotent; repeatedly deleting the same resource yields the same end state (the resource remains absent), though status codes may vary between subsequent calls (e.g., 200 OK/204 No Content versus 404 Not Found).

#Why FastAPI is Preferred Over Flask for Modern APIs


Native Asynchronous Support: FastAPI is built directly on ASGI (Starlette), providing native async/await concurrency for high-throughput, I/O-bound workloads. Flask traditionally relies on WSGI, which requires additional workarounds or gevent/gunicorn worker setups for asynchronous handling.

Automatic Data Validation & Serialization: FastAPI integrates Pydantic to automatically parse, validate, and serialize incoming requests and outgoing responses using Python type hints, eliminating manual request parsing logic.

Automatic Interactive Documentation: FastAPI generates interactive OpenAPI (Swagger UI) and ReDoc documentation out of the box directly from code declarations.

Developer Productivity & Type Safety: Autocompletion, linting, and static type analysis work seamlessly across the application stack due to standard Python type annotations.

#Question 2: Create a FastAPI application with:


● Root endpoint (/)


● /courses GET endpoint


● /courses/{course_id} using path parameter
Add:


● Query parameter level (optional)


Explain difference between path parameter and query parameter.

```python
from typing import Optional
from fastapi import FastAPI

app = FastAPI()

# Dummy database records
COURSES_DB = {
    1: {"title": "Python Basics", "level": "beginner"},
    2: {"title": "FastAPI Deep Dive", "level": "intermediate"},
    3: {"title": "Advanced Distributed Systems", "level": "advanced"},
}


@app.get("/")
def root():
    return {"message": "Welcome to the Course API"}


@app.get("/courses")
def get_courses(level: Optional[str] = None):
    if level:
        filtered = [
            course
            for course in COURSES_DB.values()
            if course["level"].lower() == level.lower()
        ]
        return {"courses": filtered}
    return {"courses": list(COURSES_DB.values())}


@app.get("/courses/{course_id}")
def get_course_by_id(course_id: int):
    course = COURSES_DB.get(course_id)
    if not course:
        return {"error": "Course not found"}
    return {"course_id": course_id, "data": course}
```

 # Q3 Add validation to:


● course_id must be integer > 0


● level must be one of: beginner,

intermediate, advanced
What happens if validation fails?

```python
from enum import Enum
from typing import Optional
from fastapi import FastAPI, HTTPException, Path, Query

app = FastAPI()


class LevelEnum(str, Enum):
    beginner = "beginner"
    intermediate = "intermediate"
    advanced = "advanced"


@app.get("/courses/{course_id}")
def get_validated_course(
    course_id: int = Path(..., gt=0, description="Course ID must be an integer > 0"),
    level: Optional[LevelEnum] = Query(
        None, description="Must be beginner, intermediate, or advanced"
    ),
):
    return {"course_id": course_id, "level": level}
```

#Authentication vs. Authorization
Authentication (AuthN): The process of verifying who a user is (e.g., verifying a username and password or an API key).

Authorization (AuthZ): The process of verifying what permissions an authenticated identity holds (e.g., determining whether a user has administrative privileges to delete a resource).

#JWT-Based Authentication
JSON Web Tokens (JWT) are self-contained tokens formatted into three parts separated by dots: Header, Payload, and Signature (header.payload.signature).

The client sends credentials (username/password) to the server.

Upon verification, the server issues a digitally signed JWT containing user claims (e.g., user_id, role, expiration timestamp) and returns it to the client.

For subsequent requests, the client transmits this token in the HTTP Authorization: Bearer <token> header.

The server validates the token's authenticity using a cryptographic secret key or public key without querying a database session store.

Why Stateless Authentication is Scalable
Because the JWT encapsulates identity claims and signature verification cryptographically within itself, the server does not need to persist session state in memory or in a centralized store like Redis. Any independent server node behind a load balancer can verify incoming requests using only the shared signing secret, allowing services to scale horizontally without session-synchronization bottlenecks.

#Why API Testing is Critical
APIs serve as public or internal communication contracts between decoupled services. Testing guarantees that endpoints conform strictly to their schemas, maintain backward compatibility, handle edge cases gracefully, and prevent functional regressions during code refactors.

#Difference Between Unit Testing and Integration Testing
Unit Testing: Tests an isolated piece of functionality (a single helper function, utility method, or validation rule) in complete isolation from external dependencies, mocking out databases and network calls.

Integration Testing: Validates that multiple components of the application function correctly together (e.g., routing, middleware, request validation, database interactions, and response formatting).

```python
from fastapi import FastAPI
from fastapi.testclient import TestClient

app = FastAPI()


@app.get("/courses/{course_id}")
def read_course(course_id: int):
    if course_id == 1:
        return {"course_id": 1, "title": "FastAPI Guide"}
    return {"error": "Not Found"}


client = TestClient(app)


def test_get_course_success():
    response = client.get("/courses/1")
    assert response.status_code == 200
    assert response.json() == {"course_id": 1, "title": "FastAPI Guide"}


def test_get_course_invalid_id():
    # Pass a string instead of an integer to trigger a 422 error
    response = client.get("/courses/invalid_id")
    assert response.status_code == 422
```

#Explanation of Typing Constructs
Optional[T]:

 Denotes that a value may either be of type T or None (equivalent to Union[T, None] or T | None). In FastAPI, it marks parameters as optional when paired with a default value of = None.

#List[T]:

Sourced from typing (or standard list[T] in Python 3.9+), it specifies that the target data structure must be a list containing elements strictly of type T.

#Union[T1, T2]:

 Declares that an acceptable value can conform to either type T1 or type T2 (e.g., Union[int, str]).

#Annotated[T, ...]:

Introduced in PEP 593, it allows attaching arbitrary metadata to a base type T without changing runtime type checking. FastAPI uses this extensively (e.g., Annotated[int, Path(gt=0)]) to combine type definitions with parameter metadata cleanly.

● Query parameter


● Request body


When should each be used?

#Comparison of Request Components
Path Parameter: Variable segments embedded inside the URL path itself (/resource/{id}).



#Query Parameter:

Key-value pairs appended after the ? delimiter in the URL (/resource?key=value).



#Request Body:

The data payload transmitted inside the body of an HTTP request (commonly formatted as JSON).

#When to Use Each



#Path Parameter  

When identifying a single, specific resource or mandatory hierarchical parent./users/42, /orders/992/items/3


#Query Parameter

When providing optional flags, filtering, sorting, searching, or paginating a collection./users?role=admin&page=2&sort=desc

#Request Body

When passing complex, structured, or sensitive data payloads for creating or updating resources.POST /users with JSON payload { "name": "Akash", "email": "a@example.com" }

#Pydantic Component Breakdown
BaseModel:

The foundational class in Pydantic used to define data schemas. Attributes declared inside a BaseModel subclass define the expected fields and types for parsing and validation.

#Field:

 A function used inside a BaseModel to provide extra configuration, validation constraints (e.g., gt, lt, min_length, regex), metadata, and OpenAPI documentation attributes for individual fields.

#validator (Pydantic v1) / field_validator (Pydantic v2):

 A method decorator used to define custom validation and transformation logic applied to a single field at a time.

#root_validator (Pydantic v1) / model_validator (Pydantic v2):

A method decorator applied to the entire model, allowing cross-validation between multiple dependent fields (e.g., ensuring end_date is strictly after start_date).

#Question 10:Use Pydantic BaseSettings to:


● Store database URL


● Store secret key


● Store debug flag

```python
!pip install pydantic-settings python-dotenv

from pydantic import Field, ConfigDict
from pydantic_settings import BaseSettings

# Create a dummy .env file for demonstration purposes
with open('.env', 'w') as f:
    f.write('DATABASE_URL="sqlite:///./test.db"\n')
    f.write('SECRET_KEY="supersecretkey"\n')
    f.write('DEBUG=True\n')


class Settings(BaseSettings):
    database_url: str = Field(env="DATABASE_URL")
    secret_key: str = Field(env="SECRET_KEY")
    debug: bool = Field(False, env="DEBUG")

    model_config = ConfigDict(env_file=".env", env_file_encoding="utf-8")


# Instantiate settings
settings = Settings()

print("Settings loaded successfully:")
print(f"Database URL: {settings.database_url}")
print(f"Secret Key: {settings.secret_key}")
print(f"Debug Mode: {settings.debug}")
```
